```python
# Importing the necessary libraries
import numpy as np
import pandas as pd
from sklearn.impute import SimpleImputer
```

```python
# Load the dataset
dataset = pd.read_csv("diabetes.csv")
X = dataset.iloc[:, : - 1].values # .values convert pandas to numpy array.
y = dataset.iloc[:, -1].values
```

```python
# Identify missing data (assumes that missing data is represented as NaN)
features = ["Glucose", "BloodPressure", "SkinThickness", "Insulin", "BMI"]
```

```python
# Print the number of missing entries in each column
for feature in features:
  zero_count = (dataset[feature] == 0).sum()
  print(feature, ":", zero_count)
```

```python
# Configure an instance of the SimpleImputer class
X[:, 1:-2][X[:, 1:-2] == 0] = np.nan # replace true location with "NaN"
# Another method
# mask = X[:, 1:-2] == 0
# X[:, 1:-2] = np.where(mask, np.nan, X[:, 1:-2])
imputer = SimpleImputer(missing_values = np.nan, strategy = "mean")
```

```python
# Fit the imputer on the DataFrame
imputer.fit(X[:, 1:-2])
```

```python
# Apply the transform to the DataFrame
X[:, 1:-2] = imputer.transform(X[:, 1:-2]) # transform actually changes the data
```

```python
#Print your updated matrix of features
print(X)
```
